```python

def variety_1d_max(x: list):
    """
    Defines variety as the sum of the max in each row of the lower triangular half of the indifference matrix
    """
    N = len(x)
    indiff = 0
    for i in range(N):
        max_indiff = 0
        for j in range(i+1, N):  
            distinguish_rad = 0
            for r in range(N//2 + 1):   # loop over all possible radii from the current bead
                if x[(i+r) % N] != x[(j+r) % N] or x[(i-r) % N] != x[(j-r) % N]:
                    break
                distinguish_rad += 1
            max_indiff = max(max_indiff, distinguish_rad)
        indiff += max_indiff
    variety = 1/indiff
    return variety
```

```python
def variety_1d_max_fast(x: list):
    """
    Defines variety as the sum of the max in each row of the lower triangular half of the indifference matrix
    """
    N = len(x)
    
    # first make a record of all the neighborhoods
    neighborhoods = defaultdict(list)  # key = neighborhood (L-R), value = list of indices possessing
    for r in range(N//2 + 1):
        for i in range(N):
            if (i+r+1) > N:
                neighborhoods[x[i-r:N] + x[0:i+r+1-N]].append(i)
            elif (i-r) < 0:
                neighborhoods[x[i-r:] + x[0:i+r+1]].append(i)
            else:
                neighborhoods[x[(i-r): (i+r+1)]].append(i)
            
    #print(neighborhoods)
    indiff = 0
    for i in range(N):
        curr_indiff = 0
        for r in range(N//2 + 1):
            s = []
            if (i+r+1) > N:
                s = neighborhoods[x[i-r:N] + x[0:i+r+1-N]]
            elif (i-r) < 0:
                s = neighborhoods[x[i-r:] + x[0:i+r+1]]
            else:
                s = neighborhoods[x[(i-r): (i+r+1)]]
            if len(s) == 1:
                curr_indiff = r
                break
        #print(curr_indiff)
        if curr_indiff == 0:
            curr_indiff = N//2 + 1
        indiff += curr_indiff
        
    variety = 1/indiff
    return variety
```
